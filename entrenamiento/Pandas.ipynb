```python
import pandas as pd
s = pd.Series(range(1,11))
print ("la serie es: ",s)
s_filtro = s[s > 5]
print ("la serie filtrada es: ",s_filtro)
suma_filtro = s_filtro.sum()
print ("la suma de los valores filtrados es: ",suma_filtro)
media_filtro = s_filtro.mean()
print ("la media de los valores filtrados es: ",media_filtro)
maximo_filtro = s_filtro.max()
print ("el valor maximo de los valores filtrados es: ",maximo_filtro)
minimo_filtro = s_filtro.min()
print ("el valor minimo de los valores filtrados es: ",minimo_filtro)
```

```python
import pandas as pd
datos = {
    "Producto": ["Arroz", "Jabon", "Arroz","Azucar", "Aceite"],
    "Cantidad": [10,7,8,2,10],
    "Precio": [1500,2000, 1500, 3000, 2500],
    "Fecha": ["2023-07-25", "2023-07-25", "2023-07-26", "2023-07-26", "2023-07-27"]
}
df = pd.DataFrame(datos)
#print(df)
df["Total"] = df["Cantidad"] * df["Precio"]
#print(df)
produc_agrup_ventas = df.groupby("Producto")["Cantidad"].sum()
print(produc_agrup_ventas)
produc_agrup_ventas = df.groupby("Producto")["Cantidad"].sum().idxmax()
print(produc_agrup_ventas)
ventas_fecha= df[df["Fecha"]== "2023-07-25"]
print(ventas_fecha)
ingresos= df.groupby("Producto")["Total"].sum()
print(ingresos)
```
