## ¿Qué es la Desviación Estándar (std)?

La **desviación estándar (std)** es una medida fundamental en estadística que nos dice cuánto se dispersan o se agrupan los valores de un conjunto de datos alrededor de su media (promedio). En otras palabras, mide la variabilidad de los datos. Un valor bajo indica que los puntos de datos tienden a estar cerca de la media, mientras que un valor alto indica que los puntos de datos están más dispersos en un rango más amplio de valores.

### Pasos para calcular la Desviación Estándar:

1.  **Calcula la media (promedio) de tus datos (μ o x̄):** Suma todos los valores y divídelos por el número total de valores.

    *   **Ejemplo:** Si tus datos son `[1, 2, 3, 4, 5]`, la media es `(1+2+3+4+5) / 5 = 15 / 5 = 3`.

2.  **Calcula la desviación de cada punto de datos con respecto a la media:** Resta la media a cada uno de tus puntos de datos.

    *   **Ejemplo:**
        *   `1 - 3 = -2`
        *   `2 - 3 = -1`
        *   `3 - 3 = 0`
        *   `4 - 3 = 1`
        *   `5 - 3 = 2`

3.  **Eleva al cuadrado cada una de esas desviaciones:** Esto se hace para eliminar los signos negativos (ya que una distancia no puede ser negativa) y para dar más peso a las desviaciones más grandes.

    *   **Ejemplo:**
        *   `(-2)^2 = 4`
        *   `(-1)^2 = 1`
        *   `(0)^2 = 0`
        *   `(1)^2 = 1`
        *   `(2)^2 = 4`

4.  **Suma todos los resultados elevados al cuadrado:** Obtendrás la suma de los cuadrados de las desviaciones.

    *   **Ejemplo:** `4 + 1 + 0 + 1 + 4 = 10`

5.  **Divide la suma de los cuadrados por el número total de puntos de datos (N) o por (N-1):**
    *   Si tienes la población completa de datos, divides por `N`. Este es el **parámetro** de desviación estándar (σ).
    *   Si tienes una muestra de los datos (que es lo más común en la práctica), divides por `N-1`. Este es el **estadístico** de desviación estándar (s). Dividir por `N-1` corrige un sesgo que subestimaría la desviación estándar de la población si solo usáramos `N` para una muestra.

    *   **Ejemplo (para una muestra):** `10 / (5 - 1) = 10 / 4 = 2.5`
    *   Este resultado intermedio es la **varianza**.

6.  **Toma la raíz cuadrada del resultado del paso 5:** Esto te devolverá la desviación estándar en las mismas unidades que tus datos originales.

    *   **Ejemplo:** `sqrt(2.5) ≈ 1.58`

### Fórmula de la Desviación Estándar:

La fórmula general para la **desviación estándar de una muestra (s)** es:

```
s = √[ Σ (xi - x̄)² / (n - 1) ]
```

Donde:
*   `s` = Desviación estándar de la muestra
*   `Σ` = Símbolo de sumatoria (significa 'sumar todo')
*   `xi` = Cada valor individual en el conjunto de datos
*   `x̄` = La media (promedio) de la muestra
*   `n` = El número total de puntos de datos en la muestra

Para la **desviación estándar de una población (σ)**, la fórmula es:

```
σ = √[ Σ (xi - μ)² / N ]
```

Donde:
*   `σ` = Desviación estándar de la población
*   `μ` = La media (promedio) de la población
*   `N` = El número total de puntos de datos en la población

En la salida de `df.describe()`, la columna `std` representa precisamente esta desviación estándar, calculada generalmente como la desviación estándar de una **muestra** (dividiendo por `n-1`), lo cual es el comportamiento predeterminado en Pandas y la mayoría de las librerías estadísticas, a menos que se especifique lo contrario.
