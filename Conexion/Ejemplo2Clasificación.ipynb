```python
#Es clasificación binaria: Manzana vs Naranja.Qué hace el modelo:
# Aprende que si la fruta pesa > 150g y es de color "naranja", es Naranja.
# Si no, es Manzana.

from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# 1. Datos de entrenamiento: ya sabemos la etiqueta
datos = pd.DataFrame({
    'peso_g': [140, 150, 160, 170, 130, 135],
    'es_naranja_color': [0, 0, 1, 1, 0, 0], # 0=rojo/verde, 1=naranja
    'tipo': ['Manzana', 'Manzana', 'Naranja', 'Naranja', 'Manzana', 'Manzana']
})

X = datos[['peso_g', 'es_naranja_color']] # Características
y = datos['tipo'] # Etiqueta a predecir

# 2. Creamos y entrenamos el modelo
modelo = DecisionTreeClassifier()
modelo.fit(X, y)

# 3. Predecimos una fruta nueva: pesa 165g y es color naranja
fruta_nueva = [[165, 1]]
prediccion = modelo.predict(fruta_nueva)

print(f"La fruta nueva es: {prediccion[0]}") # Resultado: Naranja
```
