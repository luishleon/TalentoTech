```python

from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# 1. Datos de entrenamiento: ya sabemos la etiqueta
datos = pd.DataFrame({
    'peso_g': [180, 200, 170, 220, 190],
    'longitud_cm': [15, 20, 14, 22, 16],
    'tipo': ['Pepino', 'Calabacín', 'Pepino', 'Calabacín', 'Pepino']
})

X = datos[['peso_g', 'longitud_cm']] # Características
y = datos['tipo'] # Etiqueta a predecir

# 2. Creamos y entrenamos el modelo
modelo = DecisionTreeClassifier()
modelo.fit(X, y)

# 3. Predecimos una fruta nueva: pesa 165g y es color naranja
fruta_nueva = [[200, 18]]
prediccion = modelo.predict(fruta_nueva)

print(f"La verdura es: {prediccion[0]}") # Resultado: Naranja
```
