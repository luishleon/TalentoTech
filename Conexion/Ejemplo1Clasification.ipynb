```python
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# 1. Datos de entrenamiento: ya sabemos la etiqueta
datos = pd.DataFrame({
    'horas_estudio': [2, 4, 6, 8, 5, 3],
    'asistencia': [50, 60, 80, 90, 75, 55],
    'tipo': ['Repreuba', 'Repreuba', 'Aprueba', 'Aprueba', 'Aprueba', 'Repreuba']
})

X = datos[['horas_estudio', 'asistencia']] # Características
y = datos['tipo'] # Etiqueta a predecir

# 2. Creamos y entrenamos el modelo
modelo = DecisionTreeClassifier()
modelo.fit(X, y)

# 3. Predecimos una fruta nueva: pesa 165g y es color naranja
fruta_nueva = [[200, 18]]
prediccion = modelo.predict(fruta_nueva)

print(f"La asistencia del estudiante fue: {prediccion[0]}") # Resultado: Naranja
```
